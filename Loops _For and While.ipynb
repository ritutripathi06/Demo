**Consider employee salaries corresponding to different departments**

    Category the salaries according to below criteria:
    a) salaries < =30000 : low
    b) 30000<salaries<=50000 : mid
    c) salaries>50000 : high

Display the final list that categories salaries into [low,mid,high]

```python
sal = [50000,70000,38000,19000]
# [mid,high,mid,low]
 
# Create an empty list that stores salary groups
sal_groups = []
# Run a loop on salaries
for salary in sal:
    if salary<=30000: # Condition1
        sal_groups.append("Low")
    elif 30000<salary<=50000: # Condition2
        sal_groups.append("Mid")
    elif salary>50000: # COndition3
        sal_groups.append("High")
    else: # Logic if all conditions failed
        sal_groups.append("Invalid Data")

# Come out of the loop
print(sal)
print(sal_groups)
```

```python
# for i in greeting:
#     print(i) # W
#     # e
# create a empty list
title = []
lower_data =[]
upper_data =[]
digits_data =[]
others =[]

for i in greeting.split():
    # i: Welcome
    # i: to
    # i: the...
    if i.istitle():
        # Welcome.istitle() : True
        # to.istitle() : False
        title.append(i)
    elif i.islower():
        lower_data.append(i)
    elif i.isupper():
        upper_data.append(i)
    elif i.isdigit():
        digits_data.append(i)
    else:
        others.append("Invalid data")
    # print all the lists
# comma separated method
print("Title",title,",Lower",lower_data,",Upper",upper_data,",Digits",digits_data,",Others",others)
# f string method
print(f"Title {title},Lower {lower_data},Upper {upper_data},Digits {digits_data},Others {others}")
```

**Search if employee salary 56000 is avaialble in this list.If its available then return the position,
if its not available , add salary to the list**

```python
emp_salaries = [90000,134000,45000,76000]

# run a loop on employee salaries to access each individual salary
for salary in emp_salaries:
    # check if 56000 is present in the list
    if 56000 in emp_salaries: 
        # if yes, then return its index/position
        print(emp_salaries.index(salary))
        break # stop the iteration after salary is found and its position is noted
    else:
        # if not there, add it to the list
        emp_salaries.append(56000)
        break # stop the iteration so that 56000 is added only once. # break-> it will stop loop process and cursor will come out of the loop

print(emp_salaries) # once break is executed, cursor comes here
```

**Check for the length of mismatching sublists. Whereever the lists are maintaining less size, add 0 to match with other list size**

emp_data_2 = [[101, 'raman', 3435345], [102, 'suman', 345345], [103, 'tanya'],[104,'thaman']]

```python
emp_data_2 = [[101, 'raman', 3435345], [102, 'suman', 345345], [103, 'tanya'],[104,'thaman']]
print("Before",emp_data_2)
# capture max lenggth first
lens = []
for data in emp_data_2:
    # capture length of each sublist and add it to lens list
    lens.append(len(data))
print("Data lengths",lens)
print("Max of data lengths",max(lens))

for data in emp_data_2:
    if len(data)<max(lens): # max(lens)
        # 3 < max([3,3,2,2])
        # 3<3 :False : nothing happens
        # 3<3 : False :nothing happens
        # 2<3 : True
        # 2<3 : True
        data.append(0)# will be executed when 2<3 happens

# print the empdata
print("After",emp_data_2)
```
